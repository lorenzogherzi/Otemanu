```python
import time
import math

duration = 15

print(f"CPU stress test: {duration} seconds")

start = time.time()
iterations = 0

while time.time() - start < duration:
    x = 0.0

    for i in range(100_000):
        x += math.sqrt(i) * math.sin(i)

    iterations += 1

print(f"Done. Iterations: {iterations}")
```

```python
import os
import time
import numpy as np
import psutil

process = psutil.Process(os.getpid())

print(f"Kernel PID: {process.pid}")

data = []

for step in range(10):
    # ~80 MB aggiunti a ogni iterazione
    array = np.random.random(10_000_000)
    data.append(array)

    rss = process.memory_info().rss / 1024**3

    print(
        f"Step {step + 1}/10 | "
        f"RAM kernel: {rss:.2f} GB",
        flush=True
    )

    time.sleep(1)

print("Done.")
```

```python
import os
import time
import random
import math

from tqdm import tqdm

print(f"Kernel PID: {os.getpid()}")

N = 100

results = []

for i in tqdm(
    range(N),
    desc="Survival bootstrap",
    unit="rep"
):

    # vero lavoro CPU
    value = 0.0

    for j in range(30_000):
        value += math.sqrt(j + 1)

    results.append(value)

    # durata irregolare
    time.sleep(random.uniform(0.03, 0.12))

print()
print("Analysis completed.")
print(f"Replicates: {len(results)}")
```
